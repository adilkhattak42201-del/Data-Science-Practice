<h2 style="color:#FFD700;text-align:center;">
📈 STEP 26 : Adjusted R² Score
</h2>

<p>

Adjusted R² measures the model performance while considering the number of features. It provides a more reliable evaluation than the standard R² score.

</p>

</div>

```python
# ==========================================================
# STEP 26 : Adjusted R² Score
# ==========================================================

def adj_r(r_sq, X, Y):

    return 1 - ((1-r_sq)*(len(Y)-1))/(len(Y)-X.shape[1]-1)

model = LinearRegression()

model.fit(x_train, y_train)

print(
    "Adjusted R² :",
    adj_r(model.score(x_test, y_test), x_train, y_train)
)
```
