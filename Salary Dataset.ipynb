```python
import pandas as pd  # Pandas library ko import kiya gaya hai taake CSV file parh sakein

df = pd.read_csv('salary.csv')  # Salary dataset load kiya ja raha hai

X = df.iloc[:, 1:2].values  # Level column (index 1) ko 2D array mein select kiya gaya hai
y = df.iloc[:, 2].values  # Salary column (index 2) ko target output banaya gaya hai

# Ab yahan print statements add kiye hain taake output screen par show ho:
print("--- X (Level) Output ---")
print(X)  # Employees ke levels ki values print hongi

print("\n--- y (Salary) Output ---")
print(y)  # Employees ki salaries ki values print hongi
```

```python
from sklearn.preprocessing import PolynomialFeatures  # PolynomialFeatures class ko import kiya gaya hai non-linear data ke liye
import pandas as pd  # Pandas library ko import kiya gaya hai data load karne ke liye

df = pd.read_csv('salary.csv')  # Dataset load kiya ja raha hai
X = df.iloc[:, 1:2].values  # Level column ko 2D array mein select kiya gaya hai

poly_reg = PolynomialFeatures(degree=2)  # Degree 2 ka polynomial object banaya gaya hai
X_poly = poly_reg.fit_transform(X)  # X values ko polynomial features mein convert kiya ja raha hai

print("--- X_poly Output (Polynomial Features) ---")
print(X_poly)  # Polynomial transformed values print ki ja rahi hain (bias, x, x^2)
```

<h3 style="color: #a7f3d0; margin-top: 0; font-size: 22px; border-bottom: 2px solid #34d399; padding-bottom: 8px; font-weight: 800;">💎 Step 7: Polynomial Regression Model Train Karna</h3>
<p style="color: #f0fdf4; font-size: 15px; font-weight: bold;">Is step mein hum Linear Regression model ko polynomial features (X_poly) par fit (train) karte hain taake curve line ban sakay.</p>

</div>

```python
from sklearn.linear_model import LinearRegression  # Linear regression model import kiya gaya hai
import pandas as pd  # Pandas library load ki gayi hai

df = pd.read_csv('salary.csv')  # Dataset read kiya gaya hai
X = df.iloc[:, 1:2].values  # Level column select kiya gaya hai
y = df.iloc[:, 2].values  # Salary column select kiya gaya hai

from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree=2)  # Degree 2 polynomial object banaya gaya hai
X_poly = poly_reg.fit_transform(X)  # Data ko polynomial mein transform kiya gaya hai

lin_reg_2 = LinearRegression()  # Naya linear regression object banaya gaya hai
lin_reg_2.fit(X_poly, y)  # Model ko polynomial features par train kiya ja raha hai

print("Polynomial Model Trained Successfully!")  # Model training confirmation print ki ja rahi hai
```

<h3 style="color: #dbeafe; margin-top: 0; font-size: 22px; border-bottom: 2px solid #93c5fd; padding-bottom: 8px; font-weight: 800;">💎 Step 8: Polynomial Regression Visualization aur Predictions</h3>
<p style="color: #eff6ff; font-size: 15px; font-weight: bold;">Is step mein hum Polynomial Regression ka smooth curve graph plot karte hain aur kisi naye level (jaise 6.5) ki exact salary predict karke dekhte hain.</p>

</div>

```python
import matplotlib.pyplot as plt  # Graphs plot karne ke liye matplotlib import ki gayi hai
import numpy as np  # Smooth curve ke liye numpy import kiya gaya hai
import pandas as pd  # Pandas data load karne ke liye
from sklearn.linear_model import LinearRegression  # Model import kiya gaya hai
from sklearn.preprocessing import PolynomialFeatures  # Polynomial features import ki gayi hain

df = pd.read_csv('salary.csv')  # Dataset load kiya ja raha hai
X = df.iloc[:, 1:2].values  # Level column select kiya gaya hai
y = df.iloc[:, 2].values  # Salary column select kiya gaya hai

poly_reg = PolynomialFeatures(degree=2)  # Degree 2 ka object banaya gaya hai
X_poly = poly_reg.fit_transform(X)  # Data transform kiya gaya hai

lin_reg_2 = LinearRegression()  # Model initialize kiya gaya hai
lin_reg_2.fit(X_poly, y)  # Model train kiya gaya hai

# Smooth curve ke liye X ki values ko generate karna
X_grid = np.arange(min(X), max(X), 0.1)  # 0.1 ke step se values li gayi hain smooth line ke liye
X_grid = X_grid.reshape((len(X_grid), 1))  # 2D array mein convert kiya gaya hai

# Graph plot karna
plt.scatter(X, y, color='red', label='Actual Data')  # Asli data points red dots se plot kiye gaye hain
plt.plot(X_grid, lin_reg_2.predict(poly_reg.fit_transform(X_grid)), color='blue', label='Polynomial Curve')  # Curve line plot ki gayi hai
plt.title('Salary vs Level (Polynomial Regression)', fontsize=14, fontweight='bold')  # Title set kiya gaya hai
plt.xlabel('Position Level', fontsize=12, fontweight='bold')  # X-axis label
plt.ylabel('Salary', fontsize=12, fontweight='bold')  # Y-axis label
plt.legend()  # Legend show karne ke liye
plt.grid(True, linestyle='--', alpha=0.6)  # Grid lines add ki gayi hain
plt.show()  # Graph screen par display hoga

# Level 6.5 ke liye polynomial prediction
poly_pred = lin_reg_2.predict(poly_reg.fit_transform([[6.5]]))  # 6.5 level ki prediction ki ja rahi hai
print("\nPolynomial Predicted Salary for Level 6.5:", poly_pred[0])  # Output print kiya ja raha hai
```

<h3 style="color: #f3e8ff; margin-top: 0; font-size: 22px; border-bottom: 2px solid #c084fc; padding-bottom: 8px; font-weight: 800;">💎 Step 9: Higher Degree Polynomial (Jaise Degree 4) aur Better Curve</h3>
<p style="color: #fce7f3; font-size: 15px; font-weight: bold;">Is dataset (salary.csv) ke liye final step yeh hota hai ke hum <strong>Polynomial Degree 4</strong> ka use karein, jisse curve bilkul perfect banta hai aur CEO/Manager ki outlier salary ke sath bhi model bilkul theek fit ho jata hai.</p>

</div>

```python
import matplotlib.pyplot as plt  # Graphs banane ke liye
import numpy as np  # Array operations ke liye
import pandas as pd  # Data load karne ke liye
from sklearn.linear_model import LinearRegression  # Model
from sklearn.preprocessing import PolynomialFeatures  # Polynomial features

df = pd.read_csv('salary.csv')  # Dataset read kiya ja raha hai
X = df.iloc[:, 1:2].values  # Level column
y = df.iloc[:, 2].values  # Salary column

# Degree 4 ka polynomial use karna is chote dataset ke liye best hota hai
poly_reg = PolynomialFeatures(degree=4)  # Degree 4 set ki gayi hai
X_poly = poly_reg.fit_transform(X)  # Transform data

lin_reg_4 = LinearRegression()  # Model object
lin_reg_4.fit(X_poly, y)  # Model train kiya gaya

# Smooth curve ke liye grid
X_grid = np.arange(min(X), max(X), 0.1)  # 0.1 step
X_grid = X_grid.reshape((len(X_grid), 1))  # 2D shape

# Graph plot karna
plt.scatter(X, y, color='red', label='Actual Data')  # Asli data points
plt.plot(X_grid, lin_reg_4.predict(poly_reg.fit_transform(X_grid)), color='blue', label='Degree 4 Curve')  # Smooth curve
plt.title('Salary vs Level (Polynomial Degree 4)', fontsize=14, fontweight='bold')  # Title
plt.xlabel('Position Level', fontsize=12, fontweight='bold')  # X-axis
plt.ylabel('Salary', fontsize=12, fontweight='bold')  # Y-axis
plt.legend()  # Legend
plt.grid(True, linestyle='--', alpha=0.6)  # Grid lines
plt.show()  # Display graph

# Level 6.5 ki Degree 4 ke mutabiq exact prediction
final_pred = lin_reg_4.predict(poly_reg.fit_transform([[6.5]]))  # Prediction for 6.5
print("\nFinal Degree 4 Predicted Salary for Level 6.5:", final_pred[0])  # Output print kiya ja raha hai is line mein
```

<h3 style="color: #e0f2fe; margin-top: 0; font-size: 22px; border-bottom: 2px solid #38bdf8; padding-bottom: 8px; font-weight: 800;">💎 Step 10: Model Evaluation aur Performance Metrics (R-Squared Score)</h3>
<p style="color: #f8fafc; font-size: 15px; font-weight: bold;">Is final step mein hum check karte hain ke hamara Polynomial Degree 4 model kitna accurate hai, uske liye hum <strong>R-Squared ($R^2$) Score</strong> calculate karte hain.</p>

</div>

```python
import pandas as pd  # Pandas library load ki gayi hai data parhne ke liye
from sklearn.metrics import r2_score  # R2 score calculate karne ke liye metric import kiya gaya hai
from sklearn.preprocessing import PolynomialFeatures  # Polynomial features import ki gayi hain
from sklearn.linear_model import LinearRegression  # Linear regression model import kiya gaya hai

df = pd.read_csv('salary.csv')  # Dataset load kiya ja raha hai is line mein
X = df.iloc[:, 1:2].values  # Level column select kiya gaya hai
y = df.iloc[:, 2].values  # Salary column select kiya gaya hai

poly_reg = PolynomialFeatures(degree=4)  # Degree 4 ka polynomial object banaya gaya hai
X_poly = poly_reg.fit_transform(X)  # Data ko degree 4 mein transform kiya gaya hai

lin_reg_4 = LinearRegression()  # Model object initialize kiya gaya hai
lin_reg_4.fit(X_poly, y)  # Model ko train kiya gaya hai

y_pred = lin_reg_4.predict(X_poly)  # Trained model se predictions li ja rahi hain
r2 = r2_score(y, y_pred)  # R-squared score calculate kiya ja raha hai model ki accuracy ke liye

print("R-Squared (R2) Accuracy Score:", r2)  # Model ki accuracy ka score screen par print kiya ja raha hai is line mein
```
