<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Simple Linear Regression Model</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Height aur Weight mein strong correlation aur linear relationship hai. Isay hum <b>Simple Linear Regression Model ($y = wx + b$)</b> ke zariye find karte hain, jahan aik hi predictor (X) hota hai. Model training ke doran missing parameters (Weight aur Bias) khud seekhta hai aur phir test data par perform karta hai.</p>
</div>

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Step 2: Separate Input and Output Features</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Yahan hum data ko inputs (Predictor X - Height) aur target (Output y - Weight) mein alag kar rahe hain taake model ko sikhaya ja sakay.</p>
</div>

```python
# Predictor (X) aur Target (y) ko set karna
X = data[['Height']]
print("X Shape:", X.shape)
y = data.Weight
```

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Step 3: Splitting the Data</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Data ko training aur testing sets mein taqseem karna (Yahan 30% data testing ke liye aur 70% training ke liye rakha gaya hai taake model pehle seekhe aur phir test ho).</p>
</div>

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Step 4 & 5: Model Selection & Training</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Linear Regression model ko import karke training data (X_train, y_train) par fit karte hain taake model slope ($coef\_$) aur intercept ($intercept\_$) seekh sake.</p>
</div>

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train, y_train)

# Model ke parameters check karna
print("Coefficient (W):", lr.coef_)
print("Intercept (B):", lr.intercept_)
```

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Step 6: Prediction on Training Data</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Training data par predictions generate karna taake check kar sakein ke model ne kaisa seekha hai.</p>
</div>

```python
# Model ko dubara sirf 'Height' ke sath train karna taake feature name mismatch ka error na aye
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)  # X_train yahan sirf Height wala hai

# Ab training data par predictions nikalna
y_pred_train = lr.predict(X_train)
print(y_pred_train)
```

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Step 7: Training Data Visualization & Evaluation</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Training data par actual values vs predicted values ko scatter plot se visualize karna aur MSE & R2-score ke zariye model ki performance check karna.</p>
</div>

```python
import matplotlib.pyplot as plt

plt.scatter(x=X_train, y=y_train, color='blue', label='Actual')
plt.scatter(x=X_train, y=y_pred_train, color='red', label='Predicted')
plt.legend()
plt.show()

from sklearn.metrics import mean_squared_error, r2_score

mse_train = mean_squared_error(y_train, y_pred_train)
r2_train = r2_score(y_train, y_pred_train)

display(pd.DataFrame({
    'Metric': ['MSE (Train)', 'R2 Score (Train)'],
    'Value': [mse_train, r2_train]
}).style.set_table_styles([
    {'selector': 'th', 'props': [('background-color', '#1e293b'), ('color', '#38bdf8')]},
    {'selector': 'td', 'props': [('background-color', '#0f172a'), ('color', '#f8fafc')]}
]))
```

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Step 8: Testing Data Prediction & Evaluation</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Unseen test data par predictions generate karke dekhna ke model naye data par kaisa perform kar raha hai.</p>
</div>

```python
y_pred_test = lr.predict(X_test)

plt.scatter(x=X_test, y=y_test, color='blue', label='Actual Test')
plt.scatter(x=X_test, y=y_pred_test, color='red', label='Predicted Test')
plt.legend()
plt.show()

mse_test = mean_squared_error(y_test, y_pred_test)
r2_test = r2_score(y_test, y_pred_test)

display(pd.DataFrame({
    'Metric': ['MSE (Test)', 'R2 Score (Test)'],
    'Value': [mse_test, r2_test]
}).style.set_table_styles([
    {'selector': 'th', 'props': [('background-color', '#1e293b'), ('color', '#38bdf8')]},
    {'selector': 'td', 'props': [('background-color', '#0f172a'), ('color', '#f8fafc')]}
]))
```

    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Multiple Linear Regression (Gender + Height)</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Ab hum Simple Linear Regression se agay barh kar <b>Multiple Linear Regression</b> par ja rahe hain, jahan hum aik se zyada features (Gender aur Height) istemal karke weight predict karenge.</p>
</div>

```python
# Multiple Linear Regression (Gender aur Height ke sath data split aur model fit karna)
X = data[['Gender', 'Height']]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

lr2 = LinearRegression()
lr2.fit(X_train, y_train)

print("Coefficients:", lr2.coef_)
print("Intercept:", lr2.intercept_)
```

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Multiple Linear Regression Equation & Coefficients</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Multiple linear regression ki equation $Weight = W_1(Gender) + W_2(Height) + B$ hoti hai. Yahan model ke coefficients aur intercept check karne ka code hai.</p>
</div>

```python
# Coefficients aur Intercept dobara print karna
print("Coefficients:", lr2.coef_)
print("Intercept:", lr2.intercept_)
```

<div style="background: #0f2e35; border-left: 6px solid #2dd4bf; padding: 15px 20px; border-radius: 8px; margin-bottom: 15px; border: 1px solid #134e4a;">
    <h3 style="color: #2dd4bf; margin: 0 0 5px 0; font-size: 18px;">📦 Model Prediction & Evaluation (Train & Test)</h3>
    <p style="color: #99f6e4; font-size: 14px; margin: 0;">Multiple regression model par training aur testing data ke liye predictions generate karke R2-score check karna.</p>
</div>

```python
# Multiple Regression model ka Train aur Test R2-score check karna
print("Train R2 Score:", r2_score(y_train, lr2.predict(X_train)))
print("Test R2 Score:", r2_score(y_test, lr2.predict(X_test)))
```
