```python
# Step 1: Zaroori libraries aur modules import karna
import numpy as np                          # Numerical calculations aur arrays ke liye
import matplotlib.pyplot as plt             # Graphs aur plots banane ke liye
from sklearn.datasets import fetch_california_housing # California housing dataset load karne ke liye
from sklearn.model_selection import train_test_split   # Data ko train aur test sets mein baantne ke liye
from sklearn.preprocessing import StandardScaler        # Features ko standardize (scale) karne ke liye
from sklearn.linear_model import RidgeCV     # Ridge Regression with cross-validation ke liye
from sklearn.metrics import r2_score          # Model performance/accuracy measure karne ke liye
```

        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #f8fafc;">Dataset Loading & Structure</h2>
        <p style="margin: 0; color: #94a3b8; font-size: 14px;">California housing dataset load karke dictionary format mein output dekhna</p>
    </div>
    <p style="margin: 0 0 15px 0; line-height: 1.6; color: #cbd5e1; font-size: 14px;">
        <strong>Summary:</strong> `fetch_california_housing()` function dataset ko load karta hai jismein data, target, feature names aur DESCR (description) shamil hote hain.
    </p>
    <div style="background-color: #090d16; padding: 15px; border-radius: 8px; font-family: 'Courier New', Courier, monospace; color: #38bdf8; font-size: 13px; overflow-x: auto; margin-bottom: 15px; border: 1px solid rgba(56, 189, 248, 0.3);">

```python
# Step 1: Dataset load karna aur features/target alag karna
data = fetch_california_housing()
X = data.data
y = data.target

# Yeh aapko dictionary-style output dega
print(data)
```

<div style="background: linear-gradient(135deg, #0f172a, #1e3a8a); color: #ffffff; padding: 15px; border-radius: 10px; font-family: 'Segoe UI', sans-serif;">
    <h3 style="margin: 0 0 5px 0; font-size: 16px; color: #38bdf8;">Step 2: Train-Test Split</h3>
    <p style="margin: 0; font-size: 13px; color: #cbd5e1;">Data ko 70% training aur 30% testing sets mein divide karna taaki model ko test kiya ja sake.</p>
</div>

```python
# Step 2: Data ko train aur test sets mein split karna
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Output dekhne ke liye shapes print karein
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
```

<div style="background: linear-gradient(135deg, #0f172a, #1e3a8a); color: #ffffff; padding: 25px; border-radius: 12px; margin-bottom: 20px; box-shadow: 0 8px 20px rgba(0,0,0,0.2); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif;">

        <span style="background-color: #38bdf8; color: #0f172a; padding: 4px 12px; border-radius: 15px; font-weight: bold; font-size: 13px; text-transform: uppercase;">Step 3</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #f8fafc;">Feature Scaling (StandardScaler)</h2>
        <p style="margin: 0; color: #94a3b8; font-size: 14px;">Data ko standardize karna aur scaled array output hasil karna</p>
    </div>
    <p style="margin: 0 0 15px 0; line-height: 1.6; color: #cbd5e1; font-size: 14px;">
        <strong>Summary:</strong> `StandardScaler` training data par fit hokar values ko scale karta hai taaki array output aapke bataye gaye format ki tarah mil sake.
    </p>
    <div style="background-color: #090d16; padding: 15px; border-radius: 8px; font-family: 'Courier New', Courier, monospace; color: #38bdf8; font-size: 13px; overflow-x: auto; margin-bottom: 15px; border: 1px solid rgba(56, 189, 248, 0.3);">
scaler = StandardScaler()<br>
X_train_scaled = scaler.fit_transform(X_train)<br>
X_test_scaled = scaler.transform(X_test)<br><br>
print(X_train_scaled)  # Yeh scaled array output dega
    </div>
</div>

```python
# Step 3: Feature scaling karna aur scaled array print karna
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Yeh aapko scaled array output dega jo aapne manga hai
print(X_train_scaled)
```

<div style="background: linear-gradient(135deg, #0f172a, #1e3a8a); color: #ffffff; padding: 15px; border-radius: 10px; font-family: 'Segoe UI', sans-serif;">
    <h3 style="margin: 0 0 5px 0; font-size: 16px; color: #38bdf8;">Step: RidgeCV & Linear Regression Training</h3>
    <p style="margin: 0; font-size: 13px; color: #cbd5e1;">RidgeCV se best alpha find karna aur Ridge & Linear Regression models ko train karke score check karna.</p>
</div>

```python
# RidgeCV model train karna aur best alpha ke sath score check karna
ridge_cv = RidgeCV(alphas=[0.01, 0.001, 0.1, 1.0, 10.0], cv=5)
ridge_cv.fit(X_train_scaled, y_train)
print("RidgeCV Best Score:", ridge_cv.best_score_)

# Single Ridge model train karke score nikalna
from sklearn.linear_model import Ridge
ridge_model = Ridge(alpha=0.01)
ridge_model.fit(X_train_scaled, y_train)
print("Ridge Score:", ridge_model.score(X_train_scaled, y_train))

# Normal Linear Regression model train karke score nikalna
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train_scaled, y_train)
print("Linear Regression Score:", lr.score(X_train_scaled, y_train))
```

<div style="background: linear-gradient(135deg, #0f172a, #1e3a8a); color: #ffffff; padding: 15px; border-radius: 10px; font-family: 'Segoe UI', sans-serif;">
    <h3 style="margin: 0 0 5px 0; font-size: 16px; color: #38bdf8;">Step: Categorical Encoding (OneHotEncoder & Pandas)</h3>
    <p style="margin: 0; font-size: 13px; color: #cbd5e1;">Categorical columns ko OneHotEncoder aur pd.get_dummies ke zariye numerical format mein convert karna.</p>
</div>

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd

# Dummy DataFrame assume karke OneHotEncoder apply karna
df = pd.DataFrame({'City': ['Lahore', 'Karachi', 'Islamabad']})

encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
encoder.set_output(transform="pandas")

# Fit aur transform karna categorical column par
encoded_df = encoder.fit_transform(df[['City']])
print("One-Hot Encoded DataFrame:")
print(encoded_df)

# Pandas get_dummies ka use
print("\nPandas get_dummies output:")
print(pd.get_dummies(df['City'], dtype=int))

# User input handle karna
city = "Lahore"
print("\nUser input encoded via encoder:")
print(encoder.transform([[city]]))
```
